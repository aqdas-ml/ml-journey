```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
```

```python
ds = pd.read_csv("covid_toy.csv")
ds.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test,y_train,y_test = train_test_split(ds.drop(columns=["has_covid"]), ds["has_covid"], test_size=0.2)
```

```python
si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[["fever"]])
X_test_fever = si.transform(X_test[["fever"]])
```

```python
oe = OrdinalEncoder(categories=[["Mild", "Strong"]])
X_train_cough = oe.fit_transform(X_train[["cough"]])

X_test_cough = oe.transform(X_test[["cough"]])
```

```python
ohe = OneHotEncoder(drop="first", sparse_output=False)
X_train_gender_city = ohe.fit_transform(X_train[["gender","city"]])

X_test_gender_city = ohe.transform(X_test[["gender","city"]])
```

```python
X_train_age = X_train.drop(columns=["gender","cough","city"])
X_test_age = X_test.drop(columns=["gender","cough","city"])
```

```python
X_train_trans = np.concatenate((X_train_age,X_train_gender_city,X_train_fever,X_train_cough),axis=1)
X_test_trans = np.concatenate((X_test_age,X_test_gender_city,X_test_fever,X_test_cough),axis=1)
```

```python
X_train_trans.shape
```

<h4>Instead of that method use <h2>Column Transformer</h2></h4>

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    ("trf1",SimpleImputer(),["fever"]),
    ("trf2",OrdinalEncoder(categories=[["Mild","Strong"]]),["cough"]),
    ("trf3",OneHotEncoder(drop="first",sparse_output=False),["gender","city"])
],remainder="passthrough")
```

```python
X_train_new = transformer.fit_transform(X_train)
X_test_new = transformer.transform(X_test)
```
